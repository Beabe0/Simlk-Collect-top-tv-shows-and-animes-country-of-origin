```python
import requests
import pandas as pd
import time
```

```python
r = requests.get(
    'https://data.simkl.in/discover/trending/tv/today_100.json',
    params={
        'client_id':   'your client id',
        'app-name':    'your app name',
        'app-version': '1.0',
    },
    headers={'User-Agent': 'your app name/1.0'},
)
items = r.json()
print(items[0]['title'], items[0]['watched'])
```

```python
trending_tv_26_10_30 = pd.DataFrame(items)
trending_tv_26_10_30.head()
```

```python
r = requests.get(
    'https://data.simkl.in/discover/trending/anime/today_100.json',
    params={
        'client_id':   'your client id',
        'app-name':    'your app name',
        'app-version': '1.0',
    },
    headers={'User-Agent': 'your app name/1.0'},
)
items = r.json()
print(items[0]['title'], items[0]['watched'])
```

```python
trending_anime_26_10_30 = pd.DataFrame(items)
trending_anime_26_10_30.head()
```

```python
trending_anime_26_10_30['ids'] = trending_anime_26_10_30['ids'].apply(lambda x: x.get('simkl_id') if isinstance(x, dict) else None)
trending_tv_26_10_30['ids'] = trending_tv_26_10_30['ids'].apply(lambda x: x.get('simkl_id') if isinstance(x, dict) else None)
trending_anime_26_10_30
```

```python
r = requests.get(
    f'https://api.simkl.com/anime/2729520',
    params={'client_id': 'YOUR_CLIENT_ID'},
)
show = r.json()
print(show['ratings'])
# {'simkl': {'rating': 8.6, 'votes': 11487}, 'imdb': {'rating': 8.8, 'votes': 2817264}}
```

```python
trending_anime_26_10_30['simkl_rating'] = None
trending_anime_26_10_30['simkl_votes'] = None

# Loop through each row of the DataFrame
for index, row in trending_anime_26_10_30.iterrows():
    simkl_id = row['ids']

    if pd.notna(simkl_id): # Ensure simkl_id is not NaN
        try:
            # Make API call using the simkl_id
            r = requests.get(
                f'https://api.simkl.com/anime/{int(simkl_id)}',
                params={'client_id': 'fe5f1ca38c509afbb2986bdf79874714b3303aef3514b630ebfe7cef6209903b'},
            )
            r.raise_for_status() # Raise an exception for HTTP errors
            show = r.json()

            # Extract simkl rating and votes if available
            if 'ratings' in show and 'simkl' in show['ratings']:
                trending_anime_26_10_30.loc[index, 'simkl_rating'] = show['ratings']['simkl'].get('rating')
                trending_anime_26_10_30.loc[index, 'simkl_votes'] = show['ratings']['simkl'].get('votes')

            time.sleep(0.1) # Small delay to avoid hitting rate limits

        except requests.exceptions.RequestException as e:
            print(f"API request failed for ID {simkl_id}: {e}")
        except KeyError as e:
            print(f"Key error in API response for ID {simkl_id}: {e}")
        except Exception as e:
            print(f"An unexpected error occurred for ID {simkl_id}: {e}")

# Display the updated DataFrame
trending_anime_26_10_30.head()
```

```python
trending_tv_26_10_30['simkl_rating'] = None
trending_tv_26_10_30['simkl_votes'] = None

# Loop through each row of the DataFrame
for index, row in trending_tv_26_10_30.iterrows():
    simkl_id = row['ids']

    if pd.notna(simkl_id): # Ensure simkl_id is not NaN
        try:
            # Make API call using the simkl_id
            r = requests.get(
                f'https://api.simkl.com/anime/{int(simkl_id)}',
                params={'client_id': 'fe5f1ca38c509afbb2986bdf79874714b3303aef3514b630ebfe7cef6209903b'},
            )
            r.raise_for_status() # Raise an exception for HTTP errors
            show = r.json()

            # Extract simkl rating and votes if available
            if 'ratings' in show and 'simkl' in show['ratings']:
                trending_tv_26_10_30.loc[index, 'simkl_rating'] = show['ratings']['simkl'].get('rating')
                trending_tv_26_10_30.loc[index, 'simkl_votes'] = show['ratings']['simkl'].get('votes')

            time.sleep(0.1) # Small delay to avoid hitting rate limits

        except requests.exceptions.RequestException as e:
            print(f"API request failed for ID {simkl_id}: {e}")
        except KeyError as e:
            print(f"Key error in API response for ID {simkl_id}: {e}")
        except Exception as e:
            print(f"An unexpected error occurred for ID {simkl_id}: {e}")

# Display the updated DataFrame
trending_tv_26_10_30.head()
```

```python
List_26_10_30 = pd.concat([trending_tv_26_10_30, trending_anime_26_10_30], ignore_index=True)
List_26_10_30
```

```python
List_26_10_30['country_fetched'] = None # Add a new column for fetched country

# Loop through each row of the DataFrame
for index, row in List_26_10_30.iterrows():
    # Use ids_y for anime endpoint, check for NaN
    simkl_id = row['ids']

    if pd.notna(simkl_id):
        try:
            # Make API call using the simkl_id (for anime)
            r = requests.get(
                f'https://api.simkl.com/anime/{int(simkl_id)}',
                params={'client_id': 'fe5f1ca38c509afbb2986bdf79874714b3303aef3514b630ebfe7cef6209903b'},
            )
            r.raise_for_status() # Raise an exception for HTTP errors
            show = r.json()

            # Extract country if available
            if 'country' in show:
                List_26_10_30.loc[index, 'country_fetched'] = show.get('country')

            time.sleep(0.1) # Small delay to avoid hitting rate limits

        except requests.exceptions.RequestException as e:
            print(f"API request failed for Anime ID {simkl_id}: {e}")
        except Exception as e:
            print(f"An unexpected error occurred for Anime ID {simkl_id}: {e}")

# Display the updated DataFrame
List_26_10_30.head()
```

```python
List_26_10_30 = List_26_10_30.sort_values('simkl_rating', ascending=False)
List_26_10_30
```

```python
List_26_10_30['simkl_rating'] = List_26_10_30['simkl_rating'].replace(r'\N', pd.NA)
List_26_10_30['simkl_rating'] = pd.to_numeric(List_26_10_30['simkl_rating'], errors='coerce')
print(List_26_10_30['simkl_rating'].dtypes)
```
